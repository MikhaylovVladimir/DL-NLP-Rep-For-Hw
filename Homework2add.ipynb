```python
import torch
import numpy as np
import random
from datasets import load_dataset, DatasetDict
from transformers import (
    RobertaTokenizerFast,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
    DataCollatorWithPadding
)
import evaluate
```

```python
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
random.seed(RANDOM_SEED)

dataset = load_dataset("imdb")
train_valid = dataset["train"].train_test_split(test_size=0.2, seed=RANDOM_SEED)
dataset_split = DatasetDict({
    'train': train_valid['train'],
    'val': train_valid['test'],
    'test': dataset['test']
})

tokenizer = RobertaTokenizerFast.from_pretrained("distilroberta-base")

def preprocess(example):
    return tokenizer(example['text'], truncation=True, padding='max_length', max_length=512)

tokenized_datasets = dataset_split.map(preprocess, batched=True)
tokenized_datasets = tokenized_datasets.remove_columns(['text'])
tokenized_datasets.set_format(type='torch', columns=['input_ids', 'attention_mask', 'label'])

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    acc = (predictions == labels).astype(np.float32).mean().item()
    return {"accuracy": acc}

model = AutoModelForSequenceClassification.from_pretrained("distilroberta-base", num_labels=2)

training_args = TrainingArguments(
    output_dir="./results",
    evaluation_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=3,
    weight_decay=0.01,
    seed=RANDOM_SEED,
    load_best_model_at_end=True,
    metric_for_best_model="accuracy"
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["val"],
    tokenizer=tokenizer,
    data_collator=DataCollatorWithPadding(tokenizer),
    compute_metrics=compute_metrics
)
```

```python
# Запуск обучения
trainer.train()
```

```python
from trl import AutoModelForCausalLMWithValueHead

policy_model = AutoModelForCausalLMWithValueHead.from_pretrained("lvwerra/gpt2-imdb")
```

```python
from trl import PPOTrainer, PPOConfig

config = PPOConfig(
    model_name="lvwerra/gpt2-imdb",
    learning_rate=1.41e-5,
    batch_size=3,
    mini_batch_size=1,
    ppo_epochs=2,
    log_with=None
)

ppo_trainer = PPOTrainer(
    config=config,
    model=policy_model,
    tokenizer=tokenizer
)
```

```python
import torch.nn.functional as F

prompts = ["The movie was", "I really enjoyed", "The plot felt"]

device = next(policy_model.parameters()).device

tokenized_prompts = tokenizer(prompts, return_tensors="pt", padding=True).to(device)

response_tensors = policy_model.generate(
    **tokenized_prompts,
    max_new_tokens=20,
    pad_token_id=tokenizer.eos_token_id
)

responses = tokenizer.batch_decode(response_tensors, skip_special_tokens=True)

with torch.no_grad():
    encoded = tokenizer(responses, return_tensors="pt", padding=True, truncation=True).to(reward_model.device)
    logits = reward_model(**encoded).logits
    rewards = F.softmax(logits, dim=-1)[:, 1]

queries = [tokenizer(p, return_tensors="pt").input_ids.squeeze(0).to(device) for p in prompts]
responses_ids = [tokenizer(r, return_tensors="pt").input_ids.squeeze(0).to(device) for r in responses]
```

```python
reward_list = [torch.tensor([r], device=rewards.device) for r in rewards.tolist()]
```

```python
ppo_trainer.step(queries, responses_ids, reward_list)
```

```python
import torch
import torch.nn.functional as F

prompts = ["The movie was", "I really enjoyed", "The plot felt"]
device = next(policy_model.parameters()).device

tokenized_prompts = tokenizer(prompts, return_tensors="pt", padding=True).to(device)
response_tensors = policy_model.generate(
    **tokenized_prompts,
    max_new_tokens=20,
    pad_token_id=tokenizer.eos_token_id
)

responses = tokenizer.batch_decode(response_tensors, skip_special_tokens=True)

with torch.no_grad():
    encoded = tokenizer(responses, return_tensors="pt", padding=True, truncation=True)
    encoded = {k: v.to(reward_model.device) for k, v in encoded.items()}  # 👈 перемещаем всё
    logits = reward_model(**encoded).logits
    reward_probs = F.softmax(logits, dim=-1)[:, 1]
    rewards = [torch.tensor([r], device=reward_model.device) for r in reward_probs.tolist()]

query_tensors = [tokenizer(p, return_tensors="pt").input_ids[0].to(device) for p in prompts]
response_tensors_list = [r.to(device) for r in response_tensors]

ppo_trainer.step(query_tensors, response_tensors_list, rewards)
```

```python
original_model = AutoModelForCausalLM.from_pretrained("lvwerra/gpt2-imdb").to(device)
original_responses = original_model.generate(
    **tokenizer(prompts, return_tensors="pt", padding=True).to(device),
    max_new_tokens=20,
    pad_token_id=tokenizer.eos_token_id
)
decoded_originals = tokenizer.batch_decode(original_responses, skip_special_tokens=True)

updated_responses = policy_model.generate(
    **tokenizer(prompts, return_tensors="pt", padding=True).to(device),
    max_new_tokens=20,
    pad_token_id=tokenizer.eos_token_id
)
decoded_updated = tokenizer.batch_decode(updated_responses, skip_special_tokens=True)

for i, prompt in enumerate(prompts):
    print(f"\nPrompt: {prompt}")
    print(f"Before PPO: {decoded_originals[i]}")
    print(f"After PPO : {decoded_updated[i]}")
```
